```python
import subprocess
import platform
import os

def display_title():
    """Displays a clean title banner for the network tool."""
    print("\n" + "=" * 60)
    print("     NETWORK ADMIN: PING SWEEP ACTIVE DEVICE DETECTOR")
    print("=" * 60)

def run_ping_sweep():
    """Scans a subnet range to find active devices online."""
    print("\n--- PING SWEEP SCANNER ---")
    
    # Input subnet base (e.g., 192.168.1)
    subnet = input("Enter subnet base prefix (e.g., 192.168.1): ").strip()
    if not subnet:
        print("[ERROR] Subnet prefix cannot be empty.")
        return
        
    try:
        start_range = int(input("Enter starting host number (e.g., 1): "))
        end_range = int(input("Enter ending host number (e.g., 10): "))
        
        if start_range < 1 or end_range > 254 or start_range > end_range:
            print("[ERROR] Invalid range. Host numbers must be between 1 and 254, and start <= end.")
            return
    except ValueError:
        print("\n[EXCEPTION CAUGHT] ValueError: Please enter valid integers for host ranges.")
        return

    print(f"\n[INFO] Scanning subnet {subnet}.{start_range} to {subnet}.{end_range}...")
    print("[INFO] Please wait while hosts are pinged...\n")

    active_hosts = []
    
    # Determine the correct ping command argument based on the OS
    # Windows uses '-n', Linux/macOS use '-c'
    param = '-n' if platform.system().lower() == 'windows' else '-c'
    
    # Loop through the specified host range
    for host_num in range(start_range, end_range + 1):
        ip = f"{subnet}.{host_num}"
        
        # Build command: ping -n 1 192.168.1.X (Windows) or ping -c 1 192.168.1.X (Linux/Mac)
        command = ['ping', param, '1', '-w', '500', ip] if platform.system().lower() == 'windows' else ['ping', param, '1', '-W', '1', ip]
        
        # Exception Handling for subprocess execution
        try:
            # Run the ping command quietly (suppress standard output to keep terminal clean)
            result = subprocess.run(command, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
            
            if result.returncode == 0:
                print(f"  • [ONLINE] Host active -> {ip}")
                active_hosts.append(ip)
            else:
                print(f"  • [OFFLINE / TIMEOUT] -> {ip}")
                
        except FileNotFoundError:
            print(f"[EXCEPTION CAUGHT] Ping utility executable not found on system path.")
            break
        except Exception as e:
            print(f"[EXCEPTION CAUGHT] Error scanning {ip}: {e}")

    # Output Summary
    print("\n" + "-" * 50)
    print("       PING SWEEP SCAN SUMMARY")
    print("-" * 50)
    print(f"• Total IPs Scanned : {(end_range - start_range) + 1}")
    print(f"• Active Devices Found : {len(active_hosts)}")
    if active_hosts:
        print("• Active IP Addresses  :", ", ".join(active_hosts))
    print("-" * 50)

    # File Handling: Optional save to log
    save_choice = input("\nDo you want to save the active host results to a report file? (y/n): ").strip().lower()
    if save_choice == 'y':
        filename = "active_hosts_report.txt"
        try:
            with open(filename, 'w') as file:
                file.write(f"--- Ping Sweep Report for Subnet {subnet}.x ---\n")
                for ip in active_hosts:
                    file.write(f"Active Host: {ip}\n")
            print(f"[SUCCESS] Active devices successfully saved to '{filename}'.")
        except PermissionError:
            print(f"[EXCEPTION CAUGHT] PermissionError: Cannot write to '{filename}'. Check folder permissions.")
        except IOError as io_err:
            print(f"[EXCEPTION CAUGHT] IOError: Failed to write file -> {io_err}")
        except Exception as e:
            print(f"[EXCEPTION CAUGHT] Unexpected Error -> {e}")

def main():
    """Main program controller managing options, flow, and loop controls."""
    while True:
        display_title()
        print("1. Run Ping Sweep Subnet Scan")
        print("2. Exit Program")
        
        choice = input("\nEnter your choice (1-2): ").strip()
        
        if choice == '1':
            run_ping_sweep()
        elif choice == '2':
            print("\n" + "=" * 60)
            print("  Exiting Ping Sweep Tool. Goodbye!")
            print("=" * 60 + "\n")
            break
        else:
            print("\n[ERROR] Invalid choice. Please enter 1 or 2.")
            
        # Continue or Exit prompt
        cont = input("\nWould you like to perform another scan? (y/n): ").strip().lower()
        if cont != 'y':
            print("\n" + "=" * 60)
            print("  Exiting program. Goodbye!")
            print("=" * 60 + "\n")
            break

if __name__ == "__main__":
    main()
    
```
