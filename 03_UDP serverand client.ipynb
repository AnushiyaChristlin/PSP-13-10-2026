```python
import socket

def main():
    host = '127.0.0.1'
    port = 5000
    
    while True:
        # 1. Title
        print("\n" + "=" * 50)
        print("     UDP LAB NETWORK SYSTEM (SINGLE PROGRAM)")
        print("=" * 50)
        print("1. Send Lab Alert (Client Mode)")
        print("2. Listen for Alerts (Server Mode)")
        print("3. Exit Program")
        
        # 2. Input Choice
        choice = input("\nEnter your choice (1-3): ").strip()
        
        if choice == '1':
            print("\n--- CLIENT MODE: SEND ALERT ---")
            message = input("Enter alert message to send: ").strip()
            
            if not message:
                print("[ERROR] Message cannot be empty.")
                continue
                
            client_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
            client_socket.settimeout(3.0)  # 3-second timeout for response
            
            # Exception Handling for Client
            try:
                client_socket.sendto(message.encode('utf-8'), (host, port))
                print(f"[OUTPUT] Alert successfully sent to {host}:{port}")
                
                # Receive server acknowledgment
                data, _ = client_socket.recvfrom(1024)
                print(f"[OUTPUT] Server Response -> {data.decode('utf-8')}")
                
            except socket.timeout:
                print("[ERROR] Server did not respond (Timed out). Make sure Server Mode is running.")
            except socket.error as se:
                print(f"[ERROR] Network socket error: {se}")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred: {e}")
            finally:
                client_socket.close()
                
        elif choice == '2':
            print("\n--- SERVER MODE: LISTEN FOR ALERTS ---")
            print(f"[INFO] Listening on {host}:{port} for 5 seconds...")
            
            server_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
            server_socket.settimeout(5.0)  # 5-second timeout so the listener doesn't freeze forever
            
            # Exception Handling for Server
            try:
                server_socket.bind((host, port))
                data, client_address = server_socket.recvfrom(1024)
                message = data.decode('utf-8')
                
                print(f"[OUTPUT] Alert received from {client_address}: {message}")
                
                # Send acknowledgment back to client
                response = "Acknowledgment: Alert received successfully by server."
                server_socket.sendto(response.encode('utf-8'), client_address)
                print("[OUTPUT] Acknowledgment sent back to client.")
                
            except socket.timeout:
                print("[INFO] Server timed out (No incoming messages received within 5 seconds).")
            except OSError as e:
                print(f"[ERROR] Port error: Port {port} might already be in use: {e}")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred: {e}")
            finally:
                server_socket.close()
                
        elif choice == '3':
            print("\nExiting program. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter 1, 2, or 3.")
            
        # Continue or Exit prompt
        cont = input("\nDo you want to perform another operation? (y/n): ").strip().lower()
        if cont != 'y':
            print("\nExiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
