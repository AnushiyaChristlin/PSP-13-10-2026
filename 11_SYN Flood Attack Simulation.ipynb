```python
import os
import time

def display_title():
    """Displays a clean title banner for the training module."""
    print("\n" + "=" * 60)
    print("   SYN FLOOD ATTACK SIMULATION (SECURITY AWARENESS TRAINING)")
    print("=" * 60)

def simulate_syn_flood():
    """Simulates how a SYN flood overwhelms a server's half-open connection queue."""
    print("\n--- SIMULATION: SERVER RESOURCE EXHAUSTION ---")
    
    # Configuration for the simulated server
    max_queue_size = 5  # Small queue size for easy demonstration
    syn_backlog_queue = [] # Represents half-open connections waiting for ACK
    
    print(f"[INFO] Server Initialized.")
    print(f"[INFO] TCP SYN Backlog Queue Maximum Capacity: {max_queue_size} slots.")
    print("[INFO] Normal Handshake: Client sends SYN -> Server responds SYN-ACK -> Client sends ACK.")
    print("[INFO] Attack Scenario: Attacker sends multiple SYN requests but NEVER sends the final ACK (Leaving connections 'Half-Open').\n")
    
    try:
        # Prompt user for number of malicious SYN packets to send
        num_packets = int(input("Enter number of incoming SYN requests to simulate (e.g., 8): "))
        if num_packets < 0:
            print("[ERROR] Number of packets cannot be negative.")
            return
    except ValueError:
        print("\n[EXCEPTION CAUGHT] ValueError: Please enter a valid integer for the packet count.")
        return

    print("\n[INFO] Starting simulation stream...\n")
    
    log_records = []
    server_crashed_or_blocked = False

    for i in range(1, num_packets + 1):
        packet_id = f"SYN-PACKET-{i}"
        
        # Check if queue is already full
        if len(syn_backlog_queue) >= max_queue_size:
            print(f"  • [OVERFLOW WARNING] Incoming {packet_id} dropped/ignored! SYN Backlog Queue is FULL ({len(syn_backlog_queue)}/{max_queue_size}).")
            print(f"    ⚠️  Legitimate users are now experiencing a Denial of Service (Connection Refused).")
            server_crashed_or_blocked = True
            log_records.append(f"{packet_id}: DROPPED (Queue Full - DoS Triggered)\n")
        else:
            # Add to half-open queue (simulating SYN received, waiting for ACK)
            syn_backlog_queue.append(packet_id)
            print(f"  • [STATE: HALF-OPEN] Received {packet_id}. Added to backlog queue. Current Queue: {len(syn_backlog_queue)}/{max_queue_size}")
            log_records.append(f"{packet_id}: SUCCESS (Added to Half-Open Queue)\n")
        
        time.sleep(0.4) # Slight delay for visual demonstration effect

    # Summary Report
    print("\n" + "-" * 50)
    print("       SIMULATION SUMMARY REPORT")
    print("-" * 50)
    print(f"• Total SYN Requests Processed : {num_packets}")
    print(f"• Final Backlog Queue State    : {len(syn_backlog_queue)}/{max_queue_size} slots occupied")
    if server_crashed_or_blocked:
        print("• Status : 🔴 SERVER RESOURCE EXHAUSTED (Denial of Service Achieved)")
    else:
        print("• Status : 🟢 Server handled the load within queue capacity limits.")
    print("-" * 50)

    # File Handling: Save training simulation log
    save_log = input("\nDo you want to save this simulation report to a training log file? (y/n): ").strip().lower()
    if save_log == 'y':
        filename = "syn_flood_training_report.txt"
        try:
            with open(filename, 'w') as file:
                file.writelines(log_records)
            print(f"[SUCCESS] Training log safely written to '{filename}'.")
        except PermissionError:
            print(f"[EXCEPTION CAUGHT] PermissionError: Cannot write to '{filename}'. Check folder permissions.")
        except IOError as io_err:
            print(f"[EXCEPTION CAUGHT] IOError: Failed to save file -> {io_err}")
        except Exception as e:
            print(f"[EXCEPTION CAUGHT] Unexpected Error -> {e}")

def main():
    """Main program controller managing options, flow, and loop controls."""
    while True:
        display_title()
        print("1. Run SYN Flood Resource Exhaustion Simulation")
        print("2. Exit Training Program")
        
        choice = input("\nEnter your choice (1-2): ").strip()
        
        if choice == '1':
            simulate_syn_flood()
        elif choice == '2':
            print("\n" + "=" * 60)
            print("  Exiting SYN Flood Training Module. Goodbye!")
            print("=" * 60 + "\n")
            break
        else:
            print("\n[ERROR] Invalid choice. Please enter 1 or 2.")
            
        # Continue or Exit prompt
        cont = input("\nWould you like to run another simulation? (y/n): ").strip().lower()
        if cont != 'y':
            print("\n" + "=" * 60)
            print("  Exiting program. Goodbye!")
            print("=" * 60 + "\n")
            break

if __name__ == "__main__":
    main()
```
