```python
import sys

MAX_PASSWORD_LEN = 32

def vulnerable_login():
    print("\n--- VULNERABLE LOGIN SYSTEM (UNSAFE) ---")
    print("[WARNING] In Python, standard input doesn't cause raw memory buffer overflows like C's gets(),")
    print("          but accepting unbounded input can lead to logic flaws and security risks.")
    
    # Simulating unbounded input (like gets() in C)
    password = input("Enter password (try typing more characters): ")

    if password == "secret12":
        print("[OUTPUT] Access Granted!")
    else:
        print("[ERROR] Access Denied!")

def secure_login():
    print("\n--- SECURE LOGIN SYSTEM (PATCHED) ---")
    print(f"Enter password (max {MAX_PASSWORD_LEN - 1} characters): ")

    # SECURITY FIX: Explicitly limiting length and trimming input
    password = input()[:MAX_PASSWORD_LEN - 1].strip()

    if len(password) == 0:
        print("[ERROR] Password cannot be empty.")
        return

    if password == "secret12":
        print("[OUTPUT] Success! Access Granted.")
    else:
        print("[ERROR] Access Denied! Incorrect password.")

def main():
    while True:
        print("\n==================================================")
        print("            BUFFER OVERFLOW CASE STUDY MENU         ")
        print("==================================================")
        print("1. Run Vulnerable Login (Unsafe Simulation)")
        print("2. Run Secure Login (Patched)")
        print("3. Exit Program")
        
        choice = input("\nEnter your choice (1-3): ").strip()

        if choice == '1':
            vulnerable_login()
        elif choice == '2':
            secure_login()
        elif choice == '3':
            print("\nExiting program. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter 1, 2, or 3.")
        
        cont = input("\nDo you want to perform another operation? (y/n): ").strip()
        if cont.lower() != 'y':
            print("\nExiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
