```python
import http.server
import socketserver
import socket
import os

def display_title():
    """Displays a clean title banner for the file server."""
    print("\n" + "=" * 60)
    print("     LOCAL NETWORK PROJECT FILE SERVER (HTTP)")
    print("=" * 60)

def get_local_ip():
    """Retrieves the local IP address of the machine on the network."""
    try:
        # Creates a temporary dummy socket connection to get local IP
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        s.connect(("8.8.8.8", 80))
        ip = s.getsockname()[0]
        s.close()
        return ip
    except Exception:
        return "127.0.0.1" # Fallback to localhost if offline

def list_shared_directory():
    """Displays files currently in the directory available for download."""
    print("\n--- FILES AVAILABLE IN CURRENT DIRECTORY ---")
    try:
        files = os.listdir('.')
        if not files:
            print("[INFO] The directory is empty.")
        else:
            for f in files:
                # Distinguish files vs folders
                prefix = "[DIR] " if os.path.isdir(f) else "[FILE]"
                print(f"  • {prefix} {f}")
    except Exception as e:
        print(f"[ERROR] Could not list directory contents: {e}")
    print("-" * 45)

def start_http_server():
    """Starts the HTTP server to host files for classmates on the network."""
    print("\n--- START HTTP SERVER ---")
    
    port = 8000  # Standard port for local hosting
    local_ip = get_local_ip()
    
    list_shared_directory()
    
    print(f"[INFO] Preparing to host files...")
    print(f"[INFO] Classmates on the same network can download files using:")
    print(f"       👉 http://{local_ip}:{port}\n")
    print("[INFO] Press Ctrl+C in this terminal window to stop the server when done.")
    
    handler = http.server.SimpleHTTPRequestHandler
    
    # Exception Handling for Server Startup (Port conflicts)
    try:
        with socketserver.TCPServer(("", port), handler) as httpd:
            print(f"\n[OUTPUT] Server successfully started on port {port}. Serving files...")
            httpd.serve_forever()
            
    except OSError as e:
        print(f"\n[EXCEPTION CAUGHT] Port Error: Port {port} is already in use or restricted: {e}")
        print("[INFO] Try closing other server apps or changing the port number.")
    except KeyboardInterrupt:
        print("\n\n[OUTPUT] Server stopped manually by user.")
    except Exception as e:
        print(f"\n[EXCEPTION CAUGHT] An unexpected error occurred: {e}")

def main():
    """Main program controller managing options and flow control."""
    while True:
        display_title()
        print("1. View Current Directory Files")
        print("2. Start Local HTTP Server (Share Files with Classmates)")
        print("3. Exit Program")
        
        choice = input("\nEnter your choice (1-3): ").strip()
        
        if choice == '1':
            list_shared_directory()
        elif choice == '2':
            start_http_server()
        elif choice == '3':
            print("\n" + "=" * 60)
            print("  Exiting Local File Server. Goodbye!")
            print("=" * 60 + "\n")
            break
        else:
            print("\n[ERROR] Invalid choice. Please enter a number between 1 and 3.")
            
        # Continue or Exit prompt
        cont = input("\nWould you like to return to the menu? (y/n): ").strip().lower()
        if cont != 'y':
            print("\n" + "=" * 60)
            print("  Exiting program. Goodbye!")
            print("=" * 60 + "\n")
            break

if __name__ == "__main__":
    main()
```
