```python
# Global variable shared across multiple functions
global_inventory = {
    "Apples": 50,
    "Notebooks": 120,
    "Pens": 200
}

def display_title():
    """Displays a clean title banner for the store system."""
    print("\n" + "=" * 55)
    print("     STORE MANAGEMENT SYSTEM (GLOBAL INVENTORY)")
    print("=" * 55)

def view_inventory():
    """Displays the current shared inventory using the global variable."""
    global global_inventory
    print("\n--- VIEW CURRENT INVENTORY ---")
    if not global_inventory:
        print("[INFO] The store inventory is currently empty.")
    else:
        print("Shared Inventory Status:")
        for item, count in global_inventory.items():
            print(f"  • {item:<12} : {count} units")
    print("-" * 32)

def add_stock():
    """Adds or restocks items by modifying the shared global inventory dictionary."""
    global global_inventory
    print("\n--- RESTOCK / ADD ITEMS ---")
    item = input("Enter item name to restock: ").strip().title()
    
    if not item:
        print("[ERROR] Item name cannot be blank.")
        return
        
    try:
        qty = int(input(f"Enter quantity to add for '{item}': "))
        if qty < 0:
            print("[ERROR] Quantity cannot be a negative number.")
            return
            
        # Update shared global state
        if item in global_inventory:
            global_inventory[item] += qty
        else:
            global_inventory[item] = qty
            
        print(f"\n[OUTPUT] Successfully added {qty} units.")
        print(f"Updated Stock for '{item}': {global_inventory[item]} units")
        
    except ValueError:
        print("\n[EXCEPTION CAUGHT] ValueError: Please enter a valid integer for the quantity.")
    except Exception as e:
        print(f"\n[EXCEPTION CAUGHT] Unexpected Error -> {e}")

def sell_stock():
    """Reduces stock counts from the shared global inventory dictionary."""
    global global_inventory
    print("\n--- SELL / REDUCE ITEMS ---")
    item = input("Enter item name to sell: ").strip().title()
    
    if item not in global_inventory:
        print(f"\n[ERROR] Item '{item}' does not exist in the store inventory.")
        return
        
    try:
        qty = int(input(f"Enter quantity to sell for '{item}': "))
        if qty < 0:
            print("[ERROR] Quantity cannot be a negative number.")
            return
            
        if qty > global_inventory[item]:
            print(f"\n[ERROR] Insufficient stock! Only {global_inventory[item]} units of '{item}' available.")
        else:
            # Update shared global state
            global_inventory[item] -= qty
            print(f"\n[OUTPUT] Successfully sold {qty} units of '{item}'.")
            print(f"Remaining Stock for '{item}': {global_inventory[item]} units")
            
    except ValueError:
        print("\n[EXCEPTION CAUGHT] ValueError: Please enter a valid integer for the quantity.")
    except Exception as e:
        print(f"\n[EXCEPTION CAUGHT] Unexpected Error -> {e}")

def main():
    """Main program controller managing options, flow, and loop controls."""
    while True:
        display_title()
        print("1. View Inventory Status")
        print("2. Restock / Add Items")
        print("3. Sell / Reduce Items")
        print("4. Exit Program")
        
        choice = input("\nEnter your choice (1-4): ").strip()
        
        if choice == '1':
            view_inventory()
        elif choice == '2':
            add_stock()
        elif choice == '3':
            sell_stock()
        elif choice == '4':
            print("\n" + "=" * 55)
            print("  Exiting Store Management System. Goodbye!")
            print("=" * 55 + "\n")
            break
        else:
            print("\n[ERROR] Invalid choice. Please enter a number between 1 and 4.")
            
        # Continue or Exit prompt
        cont = input("\nWould you like to perform another operation? (y/n): ").strip().lower()
        if cont != 'y':
            print("\n" + "=" * 55)
            print("  Exiting program. Goodbye!")
            print("=" * 55 + "\n")
            break

if __name__ == "__main__":
    main()
```
