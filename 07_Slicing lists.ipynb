```python
import json

def main():
    # Initial list of bank transactions
    transactions = [
        "Txn 101: -$50.00 (Grocery)",
        "Txn 102: +$1200.00 (Salary)",
        "Txn 103: -$15.20 (Coffee)",
        "Txn 104: -$120.00 (Electricity Bill)",
        "Txn 105: +$45.00 (Refund)",
        "Txn 106: -$30.00 (Gas)",
        "Txn 107: -$10.00 (Parking)"
    ]
    filename = "bank_transactions_log.json"
    
    while True:
        # 1. Title
        print("\n" + "=" * 50)
        print("       BANK TRANSACTION & SLICING MANAGER")
        print("=" * 50)
        print("1. View All Transactions")
        print("2. Add a New Transaction")
        print("3. Extract Last 5 Transactions (List Slicing)")
        print("4. Save Transactions to File (File Handling)")
        print("5. Load Transactions from File (File Handling)")
        print("6. Exit Program")
        
        # 2. Input Choice
        choice = input("\nEnter your choice (1-6): ").strip()
        
        if choice == '1':
            print("\n--- ALL BANK TRANSACTIONS ---")
            if not transactions:
                print("[INFO] No transactions recorded yet.")
            else:
                for i, txn in enumerate(transactions, 1):
                    print(f"  {i}. {txn}")
            print(f"Total Transactions: {len(transactions)}")
            
        elif choice == '2':
            print("\n--- ADD TRANSACTION ---")
            new_txn = input("Enter transaction description (e.g., Txn 108: -$25.00 (Book)): ").strip()
            
            # Exception handling for user input
            try:
                if not new_txn:
                    raise ValueError("Transaction entry cannot be blank.")
                transactions.append(new_txn)
                print(f"[OUTPUT] Successfully added transaction: '{new_txn}'")
            except ValueError as ve:
                print(f"[ERROR] {ve}")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred: {e}")
                
        elif choice == '3':
            print("\n--- EXTRACT LAST 5 TRANSACTIONS (LIST SLICING) ---")
            
            # Exception handling for list operations
            try:
                if not transactions:
                    print("[INFO] The transaction list is currently empty.")
                else:
                    # Python list slicing: [-5:] automatically handles lists with fewer than 5 items safely
                    last_five = transactions[-5:]
                    
                    print(f"\n[OUTPUT] Extracted last {len(last_five)} transaction(s):")
                    for t in last_five:
                        print(f"  • {t}")
            except Exception as e:
                print(f"[ERROR] An error occurred during slicing: {e}")
                
        elif choice == '4':
            print("\n--- SAVE TRANSACTIONS TO FILE ---")
            # Exception handling for writing files
            try:
                with open(filename, 'w') as file:
                    json.dump(transactions, file, indent=4)
                print(f"[OUTPUT] Transactions successfully saved to '{filename}'.")
            except PermissionError:
                print(f"[ERROR] Permission denied: Cannot write to '{filename}'.")
            except Exception as e:
                print(f"[ERROR] Failed to save file: {e}")
                
        elif choice == '5':
            print("\n--- LOAD TRANSACTIONS FROM FILE ---")
            # Exception handling for reading files and parsing JSON
            try:
                with open(filename, 'r') as file:
                    transactions = json.load(file)
                print(f"[OUTPUT] Successfully loaded transactions from '{filename}'.")
            except FileNotFoundError:
                print(f"[ERROR] The file '{filename}' does not exist yet.")
            except json.JSONDecodeError:
                print(f"[ERROR] The file '{filename}' is corrupted or improperly formatted.")
            except Exception as e:
                print(f"[ERROR] Failed to load file: {e}")
                
        elif choice == '6':
            print("\nExiting program. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter a number between 1 and 6.")
            
        # Continue or Exit prompt
        cont = input("\nDo you want to perform another operation? (y/n): ").strip().lower()
        if cont != 'y':
            print("\nExiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
