```python
import os
import datetime

def main():
    filename = "college_attendance_log.txt"
    
    while True:
        # 1. Title
        print("\n" + "=" * 50)
        print("       COLLEGE ATTENDANCE LOG MANAGER")
        print("=" * 50)
        print("1. Record Student Attendance (Write)")
        print("2. View Attendance Log (Read)")
        print("3. Exit Program")
        
        # 2. Input
        choice = input("\nEnter your choice (1-3): ").strip()
        
        if choice == '1':
            print("\n--- RECORD ATTENDANCE ---")
            student_name = input("Enter Student Name: ").strip()
            status = input("Enter Status (Present/Absent): ").strip().capitalize()
            
            if not student_name or status not in ["Present", "Absent"]:
                print("[ERROR] Invalid input. Name cannot be blank and status must be 'Present' or 'Absent'.")
                continue
                
            # Create a formatted log record with a timestamp
            timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
            record = f"[{timestamp}] Student: {student_name} | Status: {status}\n"
            
            # Exception handling for writing (handles locked files / permissions)
            try:
                with open(filename, 'a') as file:
                    file.write(record)
                print(f"[OUTPUT] Success! Attendance recorded for {student_name}.")
                
            except PermissionError:
                print(f"[ERROR] Could not write to '{filename}'. The file is locked or open in another application (like Excel/Notepad). Please close it and try again.")
            except OSError as e:
                print(f"[ERROR] An I/O error occurred: {e}")
                
        elif choice == '2':
            print("\n--- VIEW ATTENDANCE LOG ---")
            
            # Exception handling for reading (handles missing or locked files)
            try:
                with open(filename, 'r') as file:
                    content = file.read()
                    print(f"\n--- Log Contents ('{filename}') ---")
                    if content.strip():
                        print(content, end="")
                    else:
                        print("[INFO] The attendance log file is currently empty.")
                    print("-" * 35)
                    
            except FileNotFoundError:
                print(f"[ERROR] The attendance log file '{filename}' does not exist yet. Please record an attendance first.")
            except PermissionError:
                print(f"[ERROR] Could not read '{filename}'. The file is locked or permission is denied.")
            except OSError as e:
                print(f"[ERROR] An error occurred while reading: {e}")
                
        elif choice == '3':
            print("\nExiting program. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter 1, 2, or 3.")
            
        # Continue or Exit prompt
        cont = input("\nDo you want to perform another operation? (y/n): ").strip().lower()
        if cont != 'y':
            print("\nExiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
