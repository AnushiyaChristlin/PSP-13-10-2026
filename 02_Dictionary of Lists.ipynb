```python
import json

def main():
    # Dictionary of lists structure: {"Student Name": [mark1, mark2, ...]}
    student_marks = {}
    filename = "student_marks_log.json"
    
    while True:
        # 1. Title
        print("\n" + "=" * 50)
        print("       TEACHER STUDENT MARKS TRACKER")
        print("=" * 50)
        print("1. Add a Subject Mark for a Student")
        print("2. View All Students and Marks")
        print("3. Save Marks to File")
        print("4. Load Marks from File")
        print("5. Exit Program")
        
        # 2. Input Choice
        choice = input("\nEnter your choice (1-5): ").strip()
        
        if choice == '1':
            print("\n--- ADD SUBJECT MARK ---")
            name = input("Enter Student Name: ").strip().title()
            
            # Input validation for marks
            try:
                mark = float(input(f"Enter mark for {name}: "))
                if mark < 0 or mark > 100:
                    print("[ERROR] Marks should typically be between 0 and 100.")
                    continue
            except ValueError:
                print("[ERROR] Invalid mark format. Please enter a numerical value.")
                continue
                
            if name:
                # If student doesn't exist yet, create a new list for them
                if name not in student_marks:
                    student_marks[name] = []
                
                student_marks[name].append(mark)
                print(f"[OUTPUT] Successfully added mark {mark} for {name}.")
                print(f"Current records for {name}: {student_marks[name]}")
            else:
                print("[ERROR] Student name cannot be empty.")
                
        elif choice == '2':
            print("\n--- VIEW ALL STUDENT MARKS ---")
            if not student_marks:
                print("[INFO] No student records available yet.")
            else:
                print("\nStudent Performance List:")
                for student, marks in student_marks.items():
                    avg = sum(marks) / len(marks) if marks else 0
                    print(f"• {student} -> Marks: {marks} | Average: {avg:.2f}")
            print("-" * 35)
                    
        elif choice == '3':
            print("\n--- SAVE MARKS TO FILE ---")
            # Exception handling for writing/saving files (handles locks or permissions)
            try:
                with open(filename, 'w') as file:
                    json.dump(student_marks, file, indent=4)
                print(f"[OUTPUT] Marks successfully saved to '{filename}'.")
            except PermissionError:
                print(f"[ERROR] Permission denied: Cannot write to '{filename}'. It might be open elsewhere.")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred while saving: {e}")
                
        elif choice == '4':
            print("\n--- LOAD MARKS FROM FILE ---")
            # Exception handling for reading files (handles missing files or corrupted JSON data)
            try:
                with open(filename, 'r') as file:
                    student_marks = json.load(file)
                print(f"[OUTPUT] Successfully loaded student marks from '{filename}'.")
            except FileNotFoundError:
                print(f"[ERROR] The file '{filename}' does not exist yet. Please save data first.")
            except PermissionError:
                print(f"[ERROR] Permission denied: Cannot read '{filename}'.")
            except json.JSONDecodeError:
                print(f"[ERROR] The file '{filename}' is corrupted or improperly formatted.")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred while loading: {e}")
                
        elif choice == '5':
            print("\nExiting program. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter a number between 1 and 5.")
            
        # Continue or Exit prompt
        cont = input("\nDo you want to perform another operation? (y/n): ").strip().lower()
        if cont != 'y':
            print("\nExiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
