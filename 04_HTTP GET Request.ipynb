```python
import urllib.request
import json

def main():
    while True:
        # 1. Title
        print("\n" + "=" * 50)
        print("       PUBLIC API GET REQUEST MANAGER")
        print("=" * 50)
        print("1. Fetch Live Weather Data (Open-Meteo API)")
        print("2. Fetch Sample Public Post Data (JSONPlaceholder API)")
        print("3. Exit Program")
        
        # 2. Input Choice
        choice = input("\nEnter your choice (1-3): ").strip()
        
        if choice == '1':
            print("\n--- FETCHING LIVE WEATHER ---")
            # Free public weather API for coordinates (defaulted to Coimbatore, India)
            url = "https://api.open-meteo.com/v1/forecast?latitude=11.0168&longitude=76.9558&current_weather=true"
            
            try:
                print("[INFO] Connecting to Open-Meteo Weather API...")
                with urllib.request.urlopen(url, timeout=5) as response:
                    # Read and decode JSON response
                    data = json.loads(response.read().decode('utf-8'))
                    current = data.get('current_weather', {})
                    
                    print("\n[OUTPUT] Live Weather Results:")
                    print(f"  • Temperature : {current.get('temperature')} °C")
                    print(f"  • Wind Speed  : {current.get('windspeed')} km/h")
                    print(f"  • Wind Direction: {current.get('winddirection')}°")
                    print(f"  • Timestamp   : {current.get('time')}")
                    
            except urllib.error.HTTPError as e:
                print(f"[ERROR] HTTP Error: {e.code} - {e.reason}")
            except urllib.error.URLError as e:
                print(f"[ERROR] Network Error: Failed to reach the server. Check your internet connection. ({e.reason})")
            except json.JSONDecodeError:
                print("[ERROR] Failed to parse the response data as JSON.")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred: {e}")
                
        elif choice == '2':
            print("\n--- FETCHING SAMPLE PUBLIC DATA ---")
            # Free public testing API
            url = "https://jsonplaceholder.typicode.com/posts/1"
            
            try:
                print("[INFO] Connecting to JSONPlaceholder API...")
                with urllib.request.urlopen(url, timeout=5) as response:
                    data = json.loads(response.read().decode('utf-8'))
                    
                    print("\n[OUTPUT] Public Post Data Results:")
                    print(f"  • User ID : {data.get('userId')}")
                    print(f"  • Post ID : {data.get('id')}")
                    print(f"  • Title   : {data.get('title')}")
                    print(f"  • Body    : {data.get('body')}")
                    
            except urllib.error.HTTPError as e:
                print(f"[ERROR] HTTP Error: {e.code} - {e.reason}")
            except urllib.error.URLError as e:
                print(f"[ERROR] Network Error: Check your internet connection. ({e.reason})")
            except json.JSONDecodeError:
                print("[ERROR] Failed to parse response data as JSON.")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred: {e}")
                
        elif choice == '3':
            print("\nExiting program. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter 1, 2, or 3.")
            
        # Continue or Exit prompt
        cont = input("\nDo you want to perform another operation? (y/n): ").strip().lower()
        if cont != 'y':
            print("\nExiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
