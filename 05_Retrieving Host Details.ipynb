```python
import socket

def main():
    while True:
        # 1. Title
        print("\n" + "=" * 50)
        print("     NETWORK ADMIN: HOST DETAILS & RESOLUTION")
        print("=" * 50)
        print("1. Get Local Machine Hostname & IP")
        print("2. Resolve Remote Host/Domain (e.g., google.com)")
        print("3. Exit Program")
        
        # 2. Input Choice
        choice = input("\nEnter your choice (1-3): ").strip()
        
        if choice == '1':
            print("\n--- LOCAL MACHINE TROUBLESHOOTING ---")
            
            # Exception handling for local host retrieval
            try:
                hostname = socket.gethostname()
                ip_address = socket.gethostbyname(hostname)
                
                print(f"[OUTPUT] Hostname : {hostname}")
                print(f"[OUTPUT] IP Address : {ip_address}")
                
                # File Handling: Optional save to a troubleshooting log
                save_choice = input("\nDo you want to save these details to a log file? (y/n): ").strip().lower()
                if save_choice == 'y':
                    filename = "network_troubleshoot_log.txt"
                    try:
                        with open(filename, 'a') as file:
                            file.write(f"Hostname: {hostname} | IP Address: {ip_address}\n")
                        print(f"[SUCCESS] Details safely appended to '{filename}'.")
                    except PermissionError:
                        print(f"[ERROR] Permission denied: Cannot write to '{filename}'.")
                    except IOError as io_err:
                        print(f"[ERROR] File I/O error occurred: {io_err}")
                        
            except socket.error as se:
                print(f"[ERROR] Failed to retrieve local socket details: {se}")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred: {e}")
                
        elif choice == '2':
            print("\n--- REMOTE HOST RESOLUTION ---")
            target = input("Enter domain or hostname to resolve (e.g., github.com): ").strip()
            
            if not target:
                print("[ERROR] Target input cannot be empty.")
                continue
                
            # Exception handling for remote host resolution
            try:
                resolved_ip = socket.gethostbyname(target)
                print(f"\n[OUTPUT] Target Name : {target}")
                print(f"[OUTPUT] Resolved IP : {resolved_ip}")
                
            except socket.gaierror:
                print(f"[ERROR] DNS Resolution Failed: Could not resolve hostname '{target}'. Check your network connection or spelling.")
            except socket.error as se:
                print(f"[ERROR] Network socket error: {se}")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred: {e}")
                
        elif choice == '3':
            print("\nExiting program. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter 1, 2, or 3.")
            
        # Continue or Exit prompt
        cont = input("\nDo you want to perform another check? (y/n): ").strip().lower()
        if cont != 'y':
            print("\nExiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
