```python
import socket

def main():
    while True:
        # 1. Title
        print("\n" + "=" * 50)
        print("   SECURITY ANALYST: PORT BANNER GRABBER")
        print("=" * 50)
        print("1. Grab Service Banner from Target Port")
        print("2. Exit Program")
        
        # 2. Input Choice
        choice = input("\nEnter your choice (1-2): ").strip()
        
        if choice == '1':
            print("\n--- BANNER GRABBING SCAN ---")
            target_ip = input("Enter target IP address or hostname (e.g., 127.0.0.1): ").strip()
            
            try:
                target_port = int(input("Enter target port number (e.g., 21, 22, 80): ").strip())
            except ValueError:
                print("[ERROR] Port number must be a valid integer.")
                continue
                
            if not target_ip:
                print("[ERROR] Target IP cannot be empty.")
                continue
                
            print(f"\n[INFO] Connecting to {target_ip}:{target_port}...")
            
            # Create a TCP socket
            sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
            sock.settimeout(4.0)  # 4-second timeout to prevent hanging
            
            # Exception Handling for network and connection states
            try:
                sock.connect((target_ip, target_port))
                
                # Optional: For HTTP ports, send a quick request to trigger the banner response
                if target_port in [80, 443, 8080]:
                    http_payload = b"HEAD / HTTP/1.1\r\nHost: " + target_ip.encode() + b"\r\n\r\n"
                    sock.send(http_payload)
                
                # Attempt to receive the service banner
                banner = sock.recv(1024).decode('utf-8', errors='ignore').strip()
                
                print(f"\n[OUTPUT] Scan Results for {target_ip}:{target_port}")
                if banner:
                    print(f"  • Status : OPEN (Service Active)")
                    print(f"  • Banner :\n{banner}")
                else:
                    print(f"  • Status : OPEN, but no banner string was returned automatically.")
                    
            except socket.timeout:
                print(f"[ERROR] Connection Timed Out. Port {target_port} might be filtered by a firewall or offline.")
            except ConnectionRefusedError:
                print(f"[ERROR] Connection Refused. Port {target_port} is closed on {target_ip}.")
            except socket.gaierror:
                print(f"[ERROR] DNS Resolution Failed: Could not resolve hostname '{target_ip}'.")
            except socket.error as se:
                print(f"[ERROR] Network socket error: {se}")
            except Exception as e:
                print(f"[ERROR] An unexpected error occurred: {e}")
            finally:
                sock.close()
                
        elif choice == '2':
            print("\nExiting program. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter 1 or 2.")
            
        # Continue or Exit prompt
        cont = input("\nDo you want to scan another port? (y/n): ").strip().lower()
        if cont != 'y':
            print("\nExiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
