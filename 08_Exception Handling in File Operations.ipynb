```python
import os
import shutil

def display_title():
    """Displays a clean title banner for the program."""
    print("\n" + "=" * 60)
    print("          EXCEPTION HANDLING IN FILE OPERATIONS")
    print("=" * 60)

def safe_read_file():
    """Demonstrates handling missing or locked files during a read operation."""
    print("\n--- OPERATION 1: SAFE FILE READ ---")
    filename = input("Enter the filename to read (e.g., sample.txt): ").strip()
    
    try:
        with open(filename, 'r') as file:
            content = file.read()
            print(f"\n[OUTPUT] Successfully read '{filename}':")
            print("-" * 40)
            print(content if content.strip() else "[INFO] The file is empty.")
            print("-" * 40)
            
    except FileNotFoundError:
        print(f"\n[EXCEPTION CAUGHT] FileNotFoundError: The file '{filename}' does not exist.")
        print("[INFO] Please check the file name or create it first.")
    except PermissionError:
        print(f"\n[EXCEPTION CAUGHT] PermissionError: Access denied to read '{filename}'.")
    except OSError as e:
        print(f"\n[EXCEPTION CAUGHT] OSError: An I/O error occurred -> {e}")
    except Exception as e:
        print(f"\n[EXCEPTION CAUGHT] Unexpected Error -> {e}")

def safe_write_file():
    """Demonstrates handling permission or disk errors during a write operation."""
    print("\n--- OPERATION 2: SAFE FILE WRITE ---")
    filename = input("Enter the filename to write to (e.g., output.txt): ").strip()
    text = input("Enter text content to write: ")
    
    try:
        with open(filename, 'w') as file:
            file.write(text + "\n")
        print(f"\n[OUTPUT] Success! Content safely written to '{filename}'.")
            
    except PermissionError:
        print(f"\n[EXCEPTION CAUGHT] PermissionError: Cannot write to '{filename}'. The file might be locked or read-only.")
    except OSError as e:
        print(f"\n[EXCEPTION CAUGHT] OSError: Disk or I/O failure -> {e}")
    except Exception as e:
        print(f"\n[EXCEPTION CAUGHT] Unexpected Error -> {e}")

def safe_backup_operation():
    """Demonstrates handling missing/corrupted files in batch operations without crashing."""
    print("\n--- OPERATION 3: SAFE BATCH FILE BACKUP ---")
    
    # Pre-defined test files (includes a file that won't exist to trigger exception handling)
    files_to_backup = ["data.txt", "missing_log.txt", "config.json"]
    backup_folder = "backup_dir"
    
    try:
        os.makedirs(backup_folder, exist_ok=True)
    except Exception as e:
        print(f"[EXCEPTION CAUGHT] Could not create backup directory: {e}")
        return

    print(f"[INFO] Starting backup to folder: '{backup_folder}/'")
    success, failed = 0, 0
    
    for filename in files_to_backup:
        # Create a dummy file for testing if it doesn't exist (except the missing_log.txt)
        if filename != "missing_log.txt" and not os.path.exists(filename):
            try:
                with open(filename, 'w') as f:
                    f.write(f"Sample content for {filename}")
            except Exception:
                pass

        try:
            shutil.copy2(filename, os.path.join(backup_folder, filename))
            print(f"  • [SUCCESS] Backed up: '{filename}'")
            success += 1
            
        except FileNotFoundError:
            print(f"  • [EXCEPTION CAUGHT] FileNotFoundError: '{filename}' was missing and skipped.")
            failed += 1
        except PermissionError:
            print(f"  • [EXCEPTION CAUGHT] PermissionError: '{filename}' is locked and skipped.")
            failed += 1
        except Exception as e:
            print(f"  • [EXCEPTION CAUGHT] Error on '{filename}': {e}")
            failed += 1
            
    print(f"\n[OUTPUT] Backup Summary -> Successful: {success} | Skipped/Failed: {failed}")

def main():
    """Main program controller managing user options and flow."""
    while True:
        display_title()
        print("1. Safe Read File (Handles FileNotFoundError / PermissionError)")
        print("2. Safe Write File (Handles PermissionError / Disk Errors)")
        print("3. Safe Batch Backup (Prevents crashes on missing files)")
        print("4. Exit Program")
        
        choice = input("\nEnter your choice (1-4): ").strip()
        
        if choice == '1':
            safe_read_file()
        elif choice == '2':
            safe_write_file()
        elif choice == '3':
            safe_backup_operation()
        elif choice == '4':
            print("\n" + "=" * 60)
            print("  Exiting 'Exception Handling in File Operations'. Goodbye!")
            print("=" * 60 + "\n")
            break
        else:
            print("\n[INVALID INPUT] Please enter a valid option between 1 and 4.")
            
        # Continue or Exit prompt
        cont = input("\nWould you like to perform another operation? (y/n): ").strip().lower()
        if cont != 'y':
            print("\n" + "=" * 60)
            print("  Exiting program. Goodbye!")
            print("=" * 60 + "\n")
            break

if __name__ == "__main__":
    main()
```
